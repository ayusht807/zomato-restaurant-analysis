```python
import pandas as pd

df = pd.read_csv("data/zomato.csv")

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python
print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types and non-null values:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nUnique values in each column:")
print(df.nunique().sort_values())
```

```python
# Keep a backup of the original data
df_original=df.copy()

# Create a working copy
clean_df=df.copy()

# Remove accidental spaces around column names
clean_df.columns=clean_df.columns.str.strip()

print("Original shape:", df_original.shape)
print("Working copy shape:", clean_df.shape)
print(clean_df.columns.tolist())
```

```python
import pandas as pd

# Remove leading/trailing spaces from text columns
text_columns=clean_df.select_dtypes(include=["object","str"]).columns

for col in text_columns:
    clean_df[col]=clean_df[col].str.strip()

# Convert empty strings into proper missing values
clean_df=clean_df.replace(r"^\s*$",pd.NA,regex=True)

print("Missing values after standardization:")
print(clean_df.isnull().sum().sort_values(ascending=False))
```

```python
print("Rating values:")
print(clean_df["rate"].value_counts(dropna=False).head(20))

print("\nDistinct rating values:")
print(clean_df["rate"].dropna().unique()[:30])
```

```python
print("Cost values:")
print(
    clean_df["approx_cost(for two people)"]
    .value_counts(dropna=False)
    .head(15)
)

print("\nRestaurant types:")
print(clean_df["rest_type"].value_counts(dropna=False).head(15))

print("\nCuisine values:")
print(clean_df["cuisines"].value_counts(dropna=False).head(15))
```

```python
# Standardize and convert ratings
clean_df["rate"] = (
    clean_df["rate"]
    .astype("string")
    .str.strip()
    .str.extract(r"^(\d+(?:\.\d+)?)\s*/\s*5$", expand=False)
)

clean_df["rate"]=pd.to_numeric(clean_df["rate"], errors="coerce")

# Check the results
print("Rating data type:",clean_df["rate"].dtype)
print("Missing/unrated ratings:",clean_df["rate"].isna().sum())
print("\nRating summary:")
print(clean_df["rate"].describe())

print("\nRatings outside the expected 0–5 range:")
print(clean_df.loc[~clean_df["rate"].between(0, 5), "rate"].dropna())
```

```python
cost_col="approx_cost(for two people)"

# Remove commas and convert costs to numeric values
clean_df[cost_col]=pd.to_numeric(
    clean_df[cost_col]
    .astype("string")
    .str.replace(",", "", regex=False)
    .str.strip(),
    errors="coerce"
)

print("Cost data type:", clean_df[cost_col].dtype)
print("Missing costs:", clean_df[cost_col].isna().sum())

print("\nCost summary:")
print(clean_df[cost_col].describe())

print("\nSample cleaned costs:")
print(clean_df[cost_col].dropna().unique()[:20])
```

```python
print("Ratings:")
print(clean_df["rate"].value_counts(dropna=False).head(10))

print("\nCosts:")
print(clean_df[cost_col].value_counts(dropna=False).head(10))

print("\nDataset shape:", clean_df.shape)
```

```python
print("Dataset shape:",clean_df.shape)

print("\nMissing values:")
print(clean_df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:",clean_df.duplicated().sum())

print("\nOnline order availability:")
print(clean_df["online_order"].value_counts(dropna=False))

print("\nTable booking availability:")
print(clean_df["book_table"].value_counts(dropna=False))
```

```python
online_rating = clean_df.groupby("online_order")["rate"].agg(
    ["count", "mean", "median"]
).round(2)

print(online_rating)
```

```python
booking_rating = clean_df.groupby("book_table")["rate"].agg(
    ["count", "mean", "median"]
).round(2)

print(booking_rating)
```

```python
cuisine_counts=clean_df["cuisines"].dropna().str.split(",").explode().str.strip()

top_cuisines=cuisine_counts.value_counts().head(10)

print(top_cuisines)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(x=top_cuisines.values y=top_cuisines.index)

plt.title("Top 10 Most Common Cuisines")
plt.xlabel("Number of Listings")
plt.ylabel("Cuisine")
plt.tight_layout()

plt.savefig("images/top_cuisines.png",dpi=300,bbox_inches="tight")
plt.show()
```

```python
location_ratings=clean_df.groupby("location")["rate"].agg(
    ["count", "mean"]
)

location_ratings=location_ratings[location_ratings["count"] >= 100]

top_rated_locations = location_ratings.sort_values(
    "mean", ascending=False
).head(10).round(2)

print(top_rated_locations)
```

```python
cost_col="approx_cost(for two people)"

cost_analysis=clean_df[["rate", cost_col]].dropna().copy()

cost_analysis["cost_range"]=pd.cut(
    cost_analysis[cost_col],
    bins=[0, 300, 600, 1000, float("inf")],
    labels=["Budget (₹1–300)", "Mid-range (₹301–600)", "Premium (₹601–1000)", "Luxury (₹1000+)"],
    include_lowest=True
)

cost_summary=cost_analysis.groupby("cost_range",observed=True).agg(
    restaurant_count=("rate","count"),
    average_rating=("rate","mean"),
    median_rating=("rate","median")
).round(2)

print(cost_summary)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=cost_analysis,
    x="cost_range",
    y="rate",
    errorbar=None
)

plt.title("Average Restaurant Rating by Cost Range")
plt.xlabel("Approximate Cost for Two")
plt.ylabel("Average Rating")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig("images/ratings_by_cost.png", dpi=300, bbox_inches="tight")
plt.show()
```
